```python
import gymnasium as gym
import traci
import traci.constants as tc
import json
import pathlib

base_out_dir = "./compass_env/levels"
pathlib.Path(base_out_dir).mkdir(parents=True, exist_ok=True)
init_state_number = 1
save_case_numbers = 28
for seed in range(init_state_number):
    for case_id in range(1,save_case_numbers+1):
        env = gym.make("compass-highway-v0", render_mode=None, verbose=False, config={
            "case_num": case_id,
        })
        obs, info = env.reset(seed=seed)

        label = f"level_{case_id}_seed_{seed}"
        file_name = f"{label}.state.xml.gz"
        traci.simulation.saveState(f"{base_out_dir}/{file_name}")
        meta_path = pathlib.Path(f"{base_out_dir}/{label}.meta.json")
        meta = {
            "file_name": file_name,
            "case_id": case_id,
            "seed": seed,
            "time_bias": env.unwrapped.time_bias,
        }
        meta_path.write_text(json.dumps(meta, ensure_ascii=False, indent=2), encoding="utf-8")
```

```python
import pathlib
import random

def get_state_files(folder_path):
    """
    读取指定文件夹中所有以.state.xml.gz结尾的文件，返回完整路径列表
    
    Args:
        folder_path (str): 目标文件夹路径（相对/绝对路径均可）
    
    Returns:
        list: 符合条件的文件完整路径列表（pathlib.Path对象）
    """
    # 将路径转为Path对象，自动处理跨平台路径分隔符
    folder = pathlib.Path(folder_path)
    
    # 检查文件夹是否存在
    if not folder.is_dir():
        print(f"错误：文件夹 {folder_path} 不存在！")
        return []
    
    # 递归/非递归查找所有以.state.xml.gz结尾的文件
    # glob("*.state.xml.gz")：仅当前文件夹
    # glob("**/*.state.xml.gz")：递归子文件夹（如需递归则用这个）
    state_files = list(folder.glob("*.state.xml.gz"))
    
    # 可选：转为字符串路径（如果需要字符串而非Path对象）
    # state_files = [str(file) for file in state_files]
    
    print(f"找到 {len(state_files)} 个.state.xml.gz文件")
    return state_files

# ------------------- 调用示例 -------------------
# 1. 指定目标文件夹（替换为你的实际路径）
target_folder = "./compass_env/levels"  # 相对路径

# 2. 获取符合条件的文件列表
state_file_list = get_state_files(target_folder)

# 3. 随机选取一个文件（处理空列表情况）
if state_file_list:
    # 随机选一个
    random_file = random.choice(state_file_list)
    # 输出结果（Path对象可直接转字符串）
    print(f"随机选中的文件：{random_file}")
    # 如果需要仅文件名（不含路径）：
    print(f"仅文件名：{random_file.name}")
else:
    print("未找到任何.state.xml.gz文件！")
```
